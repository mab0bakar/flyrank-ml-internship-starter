Unit of analysis:

```python
from datasets import load_dataset
import pandas as pd

performance = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train[:10000]"
)

performance_df = pd.DataFrame(performance)

print(performance_df[
    ["report_date", "client_hash_id", "content_hash_id"]
].head())

print("Rows:", len(performance_df))
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### Limitation

This dataset contains historical performance metrics only.

It cannot directly measure:

- Content quality
- Competitor activity
- Future search demand
- Algorithm changes

The results should be treated as decision-support rather than proof of
causation.

```python
performance_df["needs_refresh"] = (
    performance_df["gsc_clicks"] <
    performance_df["gsc_clicks"].median()
).astype(int)

performance_df["leak_feature"] = performance_df["needs_refresh"]

print(
    performance_df[
        ["needs_refresh", "leak_feature"]
    ].head()
)

print(
    "Leakage example: leak_feature contains the answer itself and must be removed."
)
```
